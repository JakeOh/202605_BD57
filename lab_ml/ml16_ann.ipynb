# Imports

```python
from sklearn import datasets
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import accuracy_score

import tensorflow as tf
import keras
```

# Iris 데이터셋

```python
data, target = datasets.load_iris(return_X_y=True)
```

```python
data.shape  #> (n_samples, n_features)
```

```python
target.shape  #> (n_samples,)
```

# 이진 분류(binary classification)

*   2개의 특성만을 사용
*   0(setosa)과 0이 아닌(1-versicolor, 2-virginica) 클래스로 분류

```python
x = data[:, :2]
```

```python
x.shape
```

```python
x[:5]
```

```python
y = (target != 0).astype('int')
```

```python
y
```

# SGD Classifier(확률적 경사 하강법 분류기)

```python
# loss='log_loss': logistic 손실 함수를 최소화하는 확률적 경사 하강법을 사용하기 위해서.
# sigmoid 함수로 클래스 0이 될 확률과 클래스 1이 될 확률을 계산하기 위해서
# max_iter=10: epoch 횟수를 10번으로 제한.
# epoch: 모든 샘플을 1번 사용해서 경사하강법을 수행.
sgd = SGDClassifier(loss='log_loss', max_iter=10, n_jobs=-1, random_state=42)
```

```python
# 모델 훈련 -> 손실함수를 최소화하는 가중치(coefficients, intercepts)들을 찾는 과정.
sgd.fit(X=x, y=y)
```

```python
sgd.coef_  #> (1, 2) shape의 2차원 배열
```

```python
sgd.intercept_  #> (1,) shape의 1차원 배열
```

$z = -53.4 + 107.4 \times x_1 - 163.7 \times x_2$

```python
# SGD 모델의 예측값
y_pred = sgd.predict(X=x)
y_pred
```

```python
# SGD 모델의 정확도
accuracy_score(y_true=y, y_pred=y_pred)
```

```python
149/150
```

```python
# SGD 모델이 계산한 예측 확률
pred_prob = sgd.predict_proba(X=x)
```

```python
pred_prob.shape
```

```python
pred_prob[:5]
```

```python
pred_prob[-5:]
```

```python
pred_prob[y != y_pred]
```

# 신경망(Neural Network)

```python
# 입력층(input layer)
inputs = keras.Input(shape=(2,))
```

```python
# 출력층(output layer)
dense = keras.layers.Dense(units=1, activation='sigmoid')
```

```python
# 신경망 모델 생성
model = keras.Sequential(layers=[inputs, dense])
```

```python
# 생성된 신경망 모델 정보 요약
model.summary()
```

```python
# 신경망 모델 컴파일: optimizer(최적화 알고리즘), loss(손실 함수), metrics(평가 기준)을 설정
model.compile(optimizer=keras.optimizers.SGD(),
              loss=keras.losses.binary_crossentropy,
              metrics=[keras.metrics.binary_accuracy])
```

```python
# 신경망 모델 훈련
model.fit(x=x, y=y, epochs=10)
```

```python
# 신경망 모델이 학습을 통해서 찾은 모델 파라미터들(weights, 가중치)
model.weights
```

$z = 0.19 + 0.42 \times x_1 - 0.65 \times x_2$

```python
# 학습이 끝난 신경망 모델의 예측값 계산
y_pred = model.predict(x=x)
```

```python
y_pred.shape
```

```python
y_pred[:5]
```

```python
y_pred[-5:]
```

```python
y_pred_label = (y_pred >= 0.5).astype('int').ravel()
y_pred_label
```

```python
# 정확도
accuracy_score(y_true=y, y_pred=y_pred_label)
```

```python
model.evaluate(x=x, y=y)
```

# 은닉층(hidden layer)을 갖는 신경망

```python
inputs = keras.layers.Input(shape=(2,))  # 입력층(input layer)
dense1 = keras.layers.Dense(units=2, activation='sigmoid')  # 은닉층(hidden layer)
dense2 = keras.layers.Dense(units=1, activation='sigmoid')  # 출력층(output layer)
```

```python
# 신경망 모델 생성
model = keras.Sequential(layers=[inputs, dense1, dense2])
```

```python
model.summary()
```

```python
# 신경망 모델 컴파일
model.compile(optimizer=keras.optimizers.SGD(),
              loss=keras.losses.binary_crossentropy,
              metrics=[keras.metrics.binary_accuracy])
```

```python
# 신경망 모델 훈련
model.fit(x=x, y=y, batch_size=10, epochs=10)
```

```python
# 훈련이 끝난 신경망이 찾은 가중치들
model.weights
```

```python
model.evaluate(x=x, y=y)
```
