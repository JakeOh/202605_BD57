# EXAONE-3.5 모델

```python
pipe = transformers.pipeline(
    task='text-generation',
    model='LGAI-EXAONE/EXAONE-3.5-2.4B-Instruct',
    device=0,  # GPU 사용
    trust_remote_code=True
)
```

```python
# 메시지 템플릿: 'role'과 'content'를 키로 갖는 dict들의 list.
# role: 역할(system, user, assistant)
# content: 내용
messages = [
    {
        'role': 'system',
        'content': '너는 사용자의 질문에 정확한 답변을 해주는 친절한 AI 비서야.'
    },
    {
        'role': 'user',
        'content': '2026 아시안게임 야구 우승팀은 어느 나라야?',
    }
]
```

```python
result = pipe(messages, max_new_tokens=200)
```

pipeline 객체는 실행할 때마다 다른 답변(텍스트)을 생성함.

pipeline 파라미터:
*   `return_full_text`: 이전의 대화 기록을 모두 리턴할지 여부를 설정. 기본값은 True.
*   `do_sample`: 샘플링 전략을 사용할지 여부를 설정. 기본값은 True.
    *   `do_sample=True`: 메시지 프롬프트가 같아도 실행할 때마다 생성되는 텍스트가 달라짐.
    *   `do_sample=False`
        *   메시지 프롬프트가 같으면 항상 같은 텍스트가 생성됨.
        *   가장 확률이 높은 토큰들을 선택해서 텍스트를 생성하기 때문에.
*   샘플링 전략
    *   temperature
    *   top-k 방식
    *   top-p 방식

```python
result = pipe(messages,
              max_new_tokens=200,
              return_full_text=False,
              do_sample=False)
print(result)
```

# temperature 샘플링 전략

temperature(온도)를 1보다 큰 값으로 설정하면 AI가 텍스트를 생성할 때 선택되는 토큰들이 다양해지기 때문에 답변이 매우 무작위적이 됨.

temperature를 1보다 작은 값으로 설정하면 높은 확률을 가진 토큰들이 선택될 가능성이 더 커짐. AI가 생성하는 텍스트(답변)은 더 일관되어 짐.

```python
result = pipe(messages,
              max_new_tokens=200,
              return_full_text=False,
              do_sample=True,  # 기본값
              temperature=10.0)
print(result[0].get('generated_text'))
```

```python
result = pipe(messages,
              max_new_tokens=200,
              return_full_text=False,
              do_sample=True,  # 기본값
              temperature=0.1)
print(result[0].get('generated_text'))
```

```python
# 동전 던지기
probs = [0.5, 0.5]  # 앞면이 나올 확률, 뒷면이 나올 확률

# 실험
np.random.multinomial(n=100, pvals=probs)
```

```python
# 주사위 던지기 실험
probs = [1/6] * 6  # [1/6, 1/6, 1/6, 1/6, 1/6, 1/6]
np.random.multinomial(n=600, pvals=probs)
```

logit(로짓): softmax 함수의 아규먼트. LLM에서 어휘 사전에 포함된 각 토큰들의 점수. 점수가 높으면 선택될 확률이 높은 토큰(단어), 점수가 낮으면 선택될 확률이 낮은 토큰.

```python
logits = np.array([1, 2, 3, 4, 10])  # 5개 토큰의 점수.
probs = scipy.special.softmax(logits)
print(probs)
```

```python
np.random.multinomial(n=500, pvals=probs)
```

```python
logits_scaled = logits / 10
print('logits_scaled:', logits_scaled)

probs = scipy.special.softmax(logits_scaled)
print('probabilities:', probs)

ex = np.random.multinomial(n=500, pvals=probs)
print('실험 결과:', ex)
```

```python
logits_scaled = logits / 0.1
print('logits_scaled:', logits_scaled)

probs = scipy.special.softmax(logits_scaled)
print('probabilities:', probs)

ex = np.random.multinomial(n=500, pvals=probs)
print('실험 결과:', ex)
```


*   LLM에서 온도(temperature)는 logit 값들을 나눠주는 수.
*   logit 값들을 1보다 큰 수로 나눈 후 softmax 함수로 확률을 계산하면, logit 값이 작은 토큰들의 확률이 커지는 효과.
    *   토큰을 선택하는 다양성이 커짐.
    *   생성되는 텍스트가 다양해짐.
*   logit 값을 1보다 작은 수로 나눈 후 softmax 함수로 확률을 계산하면, logit 값이 큰 토큰들의 확률을 더 크게 만드는 효과
    *   토큰을 선택하는 다양성이 작아짐.
    *   거의 비슷한 텍스트들이 생성됨.
